Expects `/kaggle/input/tomato-thrips-enose/e_nose_thrips_60rep_per_treat_time.csv` locally or repo root.

```python
import pandas as pd, os
P = '/kaggle/input/datasets/suhanithakur00/tomato-thrips-enose/e_nose_thrips_60rep_per_treat_time.csv'
P = P if os.path.exists(P) else 'e_nose_thrips_60rep_per_treat_time.csv'
df = pd.read_csv(P, usecols=['SampleID','Treatment','Time_h','Replicate','Phase'])
print('rows', len(df), '| samples', df.SampleID.nunique())
print(df.Treatment.unique(), sorted(df.Time_h.unique()))
print(df.groupby('Phase').size())
print('plants', (df.Treatment + '_' + df.Replicate.astype(str)).nunique())
```
